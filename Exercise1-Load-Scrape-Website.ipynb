```python
## README FIRST
## This file describes how to download a web page. The goal is to understand how Beautiful Soup fetches web content and displays it as text.
## Change the url to get a different web page

import requests
from bs4 import BeautifulSoup

url = "https://www.nvidia.com"
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')

# Extract all text
text = soup.get_text()

# Print or save the text
print(text)

# Optionally, save to a file
with open('apple_text.txt', 'w', encoding='utf-8') as f:
    f.write(text)
```
