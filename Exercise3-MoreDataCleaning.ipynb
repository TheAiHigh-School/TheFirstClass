```python
import requests
from bs4 import BeautifulSoup
import re

response = requests.get("https://www.apple.com")
response.raise_for_status()

soup = BeautifulSoup(response.content, "html.parser")

all_text = soup.get_text(strip=True)

# Remove empty lines and split into lines
lines = (line.strip() for line in all_text.splitlines())  # Generator for efficiency
lines = list(line for line in lines if line)  # Remove empty strings and convert to list

print("All Text (Cleaned):\n")
for line in lines:
    print(line,"\n")

words = re.findall(r'\b\w+\b', all_text.lower())
word_count = len(words)
print("\nWord Count:", word_count)
```
